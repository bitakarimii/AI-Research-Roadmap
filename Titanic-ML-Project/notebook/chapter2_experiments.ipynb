```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.hist(train["Age"].dropna(), bins=30)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(train["Fare"], bins=30)

plt.title("Fare Distribution")
plt.xlabel("Fare")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
train["Pclass"].value_counts().sort_index().plot(kind="bar")

plt.title("Passengers by Ticket Class")
plt.xlabel("Passenger Class")
plt.ylabel("Count")

plt.show()
```

```python
train["Sex"].value_counts().plot(kind="bar")

plt.title("Passenger Sex")
plt.xlabel("Sex")
plt.ylabel("Count")

plt.show()
```
